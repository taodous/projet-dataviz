# Exploration of the 2026 survey: AI theme, France vs. EU

Run from the repository root environment (`pip install -r requirements.txt`). Outputs are cleared before commit.

```python
import pandas as pd
from so_ai.data import load_survey, load_schema, question_text
from so_ai import population as pop

df = load_survey()
schema = load_schema()
df.shape
```

## Who answered? Respondents per zone (country question comes late: many missing)

```python
df['Zone'].value_counts()
```

## Non-missing answers of the AI questions, in France

```python
ai_cols = ['AISelect','AIFreq_1','AISentiment','AITrust','AIComplex','AIUse','AIAvoid','LearnAI','AISat_2']
fr = df[df.Zone == pop.FRANCE]
pd.DataFrame({'France': fr[ai_cols].notna().sum(), 'All': df[ai_cols].notna().sum()})
```

## Dashboard indicators by zone

```python
from so_ai.dashboard import indicators_by_zone
indicators_by_zone(df).pivot(index='indicator', columns='zone', values='share').round(3)
```

## Reasons to avoid AI, by zone

```python
from so_ai.dashboard import avoid_reasons_by_zone
avoid_reasons_by_zone(df).pivot(index='answer', columns='group', values='share').round(3)
```
